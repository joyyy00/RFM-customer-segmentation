Cleaning

```python
# Create a copy 
df = df_raw.copy()
```

```python
# Standardize Columns 
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
df = df.rename(columns={"invoicedate": "invoice_date", "stockcode": "stock_code"})
```

```python
# Drop rows with missing customer id
df = df.dropna(subset=['customer_id'])
```

```python
# Filter out rows with negative quantity
df = df[df['quantity'] > 0]
```

```python
# Filter our rows with negative or 0 price
df = df[df['price'] > 0]
```

```python
# Fix data types
# customer_id from float to string
df.loc[:,'customer_id'] = df['customer_id'].astype(int).astype(str)
```

```python
# Create total price column
df.loc[:, 'total_price'] = df['quantity'] * df['price'] 
```

```python
df.head()
```

```python
# Check for irregular stock codes 
import re
weird_codes = df[df['stock_code'].astype(str).str.contains(r'^[A-Za-z]+$')]['stock_code'].unique()
print(weird_codes)
```

```python
# Filter out irregular stock codes
irregular_codes = ['POST' 'M' 'PADS' 'ADJUST' 'D' 'DOT']
df = df[~df['stock_code'].isin(irregular_codes)]
```
